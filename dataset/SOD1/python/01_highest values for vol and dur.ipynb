```python
import pandas as pd
import os

# 1. Load the dataset
# Replace with your actual file name if it differs
df = pd.read_csv(r"C:\Users\oozgen\Desktop\phd\dataset\SOD1_main.csv")

# 2. Sort the data 
# We sort the grouping columns normally (True), but sort the volume and period 
# in descending order (False) so the largest numbers appear first.
df_sorted = df.sort_values(
    by=['ISIS', 'Sequence', 'Location', 'Smiles', 'ASO_volume(nM)', 'Treatment_Period(hours)'],
    ascending=[True, True, True, True, False, False]
)

# 3. Drop the duplicates
# By keeping the 'first' row, we automatically keep the one with the 
# highest ASO_volume and Treatment_Period because of how we sorted it above.
df_deduplicated = df_sorted.drop_duplicates(
    subset=['ISIS', 'Sequence', 'Location', 'Smiles'],
    keep='first'
)

# 4. Save the cleaned dataset to a new CSV file
df_deduplicated.to_csv(r"C:\Users\oozgen\Desktop\phd\dataset\SOD1_cleaned_v1.csv", index=False)

# Optional: Print the results to see how many rows were removed
print(f"Original rows: {len(df)}")
print(f"Remaining rows: {len(df_deduplicated)}")
```
