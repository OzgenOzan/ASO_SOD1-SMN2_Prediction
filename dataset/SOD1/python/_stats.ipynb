```python
import pandas as pd
import os

df = pd.read_csv(r"C:\Users\oozgen\Desktop\phd\dataset\SOD1_cleaned_v2.csv")

# Display the first 5 rows to ensure it loaded correctly
display(df.head())
```

```python
# Check total rows, columns, data types, and memory usage
print("--- DataFrame Info ---")
df.info()

print("\n--- Missing Values per Column ---")
# This tells you if any columns have blank/NaN values
print(df.isnull().sum())
```

```python
# 1. Statistics for NUMERIC columns 
# (e.g., Density, ASO_volume, Treatment_Period, Inhibition, seq_length)
print("--- Numeric Statistics ---")
display(df.describe())

# 2. Statistics for CATEGORICAL/TEXT columns 
# (e.g., Target_gene, Cell_line, Sequence, Smiles)
print("--- Categorical Statistics ---")
display(df.describe(include=['object', "str"]))
```

```python
# See the COUNT of unique values for every single column
print("--- Number of Unique Values per Column ---")
display(df.nunique())

# --- Deep Dive into Specific Columns ---

# View the actual unique names of the cell lines used
print("\nUnique Cell Lines:")
print(df['Cell_line'].unique())

# View the top 10 most frequently targeted genes
print("\nTop 10 Target Genes by frequency:")
print(df['Target_gene'].value_counts().head(10))

# View the different types of modifications applied
print("\nUnique Modifications:")
print(df['Modification'].value_counts())
```

```python
# Average inhibition percentage per Target Gene
avg_inhibition_by_gene = df.groupby('Target_gene')['Inhibition(%)'].mean().sort_values(ascending=False)

print("--- Average Inhibition (%) by Target Gene ---")
display(avg_inhibition_by_gene.head(10)) # Shows the top 10 most inhibited genes on average
```

```python
# 1. Count how many negative values you actually have
negative_count = len(df[df['Inhibition(%)'] < 0])
total_rows = len(df)
print(f"There are {negative_count} rows with negative inhibition out of {total_rows}.")

# 2. Look at the statistics of ONLY the negative values
print("\n--- Stats for Negative Inhibition Values ---")
display(df[df['Inhibition(%)'] < 0]['Inhibition(%)'].describe())

# 3. View the top 30 most extremely negative rows to see if a specific gene or ASO is responsible
extreme_negatives = df.sort_values(by='Inhibition(%)', ascending=True).head(30)
display(extreme_negatives[['Target_gene', 'Sequence', 'Inhibition(%)']])
```
